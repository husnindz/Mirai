## Feature Engineering(For Dashboard)

```python
dokter_feb['bulan'] = 'Februari'
dokter_maret['bulan'] = 'Maret'
dokter_april['bulan'] = 'April'
```

## Merge Table

```python
df_dokter_all = pd.concat(
    [dokter_feb, dokter_maret, dokter_april],
    ignore_index=True
)
```

```python
df_dokter_all.info()
```

```python
df_dokter_all.duplicated().sum()
```

```python
df_dokter_all.to_csv('dokter_all.csv', index=False)
```

# Cleaning Data

```python
df_dokter_all['NM_UNIT'].unique()
```

```python
selected_unit = [
    'SPES. PENYAKIT DALAM',
    'SPES. PARU PARU',
    'JANTUNG DAN PEMBULUH DARAH'
]

df_dokter_filtered = df_dokter_all[
    df_dokter_all['NM_UNIT'].isin(selected_unit)
]
```

```python
df_dokter_filtered = df_dokter_filtered.drop(columns=['NAMA'])
```

```python
df_dokter_filtered.info()
```

Cek Jumlah Pasien yang periksa lebih dari 1 kali

```python
(df_dokter_filtered['NO_REG'].value_counts() > 1).sum()
```

Cek Jumlah Pasien yang periksa ke berbagai unit

```python
df_dokter_filtered.groupby('NO_REG')['NM_UNIT'].nunique().value_counts()
```

Hapus yang tidak ada diagnosa penyakit

```python
df_dokter_filtered = df_dokter_filtered[df_dokter_filtered['NM_PENYAKIT'].notna()]
```

```python
df_dokter_filtered.info()
```

```python
df_dokter_filtered.to_csv('dokter_filtered.csv', index=False)
```
