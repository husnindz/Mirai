# Merge Data

```python
df_lab['NO_REG'] = df_lab['NO_REG'].astype(int)
df_dokter['NO_REG'] = df_dokter['NO_REG'].astype(int)
```

```python
df_lab['HAS_LAB'] = 1
df_dokter['HAS_DOCTOR'] = 1
```

```python
df_merge = pd.merge(
    df_dokter,
    df_lab,
    on='NO_REG',
    how='outer'
)
```

```python
df_merge['HAS_LAB'] = df_merge['HAS_LAB'].fillna(0)
df_merge['HAS_DOCTOR'] = df_merge['HAS_DOCTOR'].fillna(0)
```

# Feature Selection

```python
selected_features = [
    'NO_REG',
    'JENIS_KELAMIN',
    'UMUR_TAHUN',
    'cholesterol total',
    'creatinin',
    'glukosa darah puasa',
    'glukosa darah sewaktu',
    'hgb',
    'lymfosit%',
    'mch',
    'mchc',
    'mcv',
    'ureum',
    'wbc',
    'NM_UNIT'
]
```

```python
df_clean = df_merge[selected_features].copy()
```

```python
df_clean = df_clean.rename(columns={
    'glukosa darah puasa': 'fbs',
    'glukosa darah sewaktu': 'rbs'
})
```

```python
df_clean.info()
```

# Data Transformation

```python
# Encode gender
gender_encoder = LabelEncoder()
df_clean['JENIS_KELAMIN'] = gender_encoder.fit_transform(df_clean['JENIS_KELAMIN'])
```

```python
# Mapping target
mapping_nm_unit = {
    'JANTUNG DAN PEMBULUH DARAH': 1,
    'SPES. PENYAKIT DALAM': 2,
    'SPES. PARU PARU': 3
}

df_clean['NM_UNIT'] = df_clean['NM_UNIT'].map(mapping_nm_unit)

# Isi NaN target menjadi 0
df_clean['NM_UNIT'] = df_clean['NM_UNIT'].fillna(0)
```

```python
# Drop kolom yang tidak dipakai
df_clean = df_clean.drop(columns=['NO_REG'])
```
