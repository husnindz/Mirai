## Import Semua Packages/Library yang Digunakan

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data Loading

```python

dokter = pd.read_csv('/content/dokter_all.csv')
lab = pd.read_csv('/content/lab_pivot_all.csv')

print("DATA DOKTER")
display(dokter.head())

print("\nDATA LAB")
display(lab.head())
```

## Assesing Data

```python
print("=== DOKTER ===")
dokter.info()

print("\n=== LAB ===")
lab.info()

print("\nMissing Value Dokter")
display(dokter.isna().sum())

print("\nMissing Value Lab")
display(lab.isna().sum())
```

## Data Cleaning

### hapus duplikat

```python
dokter = dokter.drop_duplicates()

lab = lab.drop_duplicates()
```

###isi missing numerik

```python
for col in dokter.select_dtypes(include='number'):
    dokter[col].fillna(dokter[col].median(), inplace=True)

for col in lab.select_dtypes(include='number'):
    lab[col].fillna(lab[col].median(), inplace=True)
```

### isi kategori

```python
for col in dokter.select_dtypes(include='object'):
    dokter[col].fillna("Unknown", inplace=True)

for col in lab.select_dtypes(include='object'):
    lab[col].fillna("Unknown", inplace=True)

print("Cleaning selesai")
```

## Exploratory Data Analysis (EDA)

### Statistik Deskriptif

```python
display(
    lab.describe()
)
```

#UNIVARIATE ANALYSIS

### Analisis Distribusi Variabel Numerik

```python
numeric_cols = lab.select_dtypes(
    include='number'
)

top_numeric = (
    numeric_cols.var()
    .sort_values(ascending=False)
    .head(5)
    .index
)

for col in top_numeric:

    plt.figure(figsize=(8,4))

    sns.histplot(
        lab[col],
        bins=30,
        kde=True
    )

    plt.title(
        f'Distribusi {col}'
    )

    plt.xlabel(col)

    plt.show()
```

#### Insight Univariate
- Distribusi data menunjukkan bahwa beberapa parameter laboratorium memiliki persebaran yang cukup luas. Hal ini menunjukkan adanya variasi nilai antar pasien pada beberapa parameter tertentu.

### Analisis Outliner

```python
for col in top_numeric:

    plt.figure(figsize=(8,4))

    sns.boxplot(
        x=lab[col]
    )

    plt.title(
        f'Boxplot {col}'
    )

    plt.show()
```

### Insight Outlier
Beberapa parameter laboratorium menunjukkan adanya outlier, yang menandakan terdapat nilai ekstrem pada beberapa pasien.

# Bivariate Analysis

```python
corr_matrix = numeric_cols.corr()

corr_pairs = (
    corr_matrix.unstack()
    .sort_values(ascending=False)
)

corr_pairs = corr_pairs[
    corr_pairs < 1
]

top_pair = corr_pairs.index[0]

var1 = top_pair[0]
var2 = top_pair[1]

plt.figure(figsize=(8,6))

sns.scatterplot(
    x=lab[var1],
    y=lab[var2]
)

plt.title(
    f'Scatterplot {var1} vs {var2}'
)

plt.xlabel(var1)
plt.ylabel(var2)

plt.show()
```

### Insight Bivariate
- Scatterplot menunjukkan adanya hubungan antar dua parameter laboratorium dengan korelasi tertinggi. Pola hubungan ini dapat membantu memahami keterkaitan antar parameter medis.

## Multivariate Analysis

```python
top10_cols = (
    numeric_cols.var()
    .sort_values(ascending=False)
    .head(10)
    .index
)

plt.figure(figsize=(12,8))

sns.heatmap(
    lab[top10_cols].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title(
    'Heatmap Top 10 Variabel'
)

plt.show()
```

```python
sns.pairplot(
    lab[top10_cols[:4]]
)

plt.show()
```

### Insight Multivariate
-Analisis multivariat menunjukkan hubungan antar beberapa parameter laboratorium secara bersamaan. Heatmap dan pairplot membantu melihat pola korelasi serta distribusi antar variabel penting dalam dataset.

## Visualization & Explanatory Analysis

### Parameter laboratorium yang paling dominan

```python

jumlah_data = (
    numeric_cols.count()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=jumlah_data.values,
    y=jumlah_data.index
)

plt.title(
    '10 Parameter Laboratorium yang Paling Sering Diperiksa'
)

plt.xlabel('Jumlah Data')
plt.ylabel('Parameter')

plt.show()
```

### Insight
- Parameter laboratorium dengan nilai rata-rata tertinggi menunjukkan variabel yang paling dominan pada dataset. Variabel tersebut dapat digunakan sebagai indikator utama dalam analisis data pasien.

### Korelasi signifikan parameter laboratorium

```python
corr_matrix = numeric_cols.corr()

corr_pairs = (
    corr_matrix.unstack()
    .sort_values(ascending=False)
)

corr_pairs = corr_pairs[
    corr_pairs < 1
]

top_corr = corr_pairs.head(10)

top_corr_df = pd.DataFrame(
    top_corr
).reset_index()

top_corr_df.columns = [
    'Variabel 1',
    'Variabel 2',
    'Nilai Korelasi'
]

display(top_corr_df)
```
